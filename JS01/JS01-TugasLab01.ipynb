> Jawab:

> Dampak Lingkungan dan Energi:
- Emisi Karbon yang Ekstrem: Melatih model AI modern membutuhkan daya
komputasi masif yang beroperasi berminggu-minggu, sehingga mengonsumsi listrik
dalam jumlah raksasa. Sebagai contoh, melatih satu model AI berskala besar dapat
memproduksi emisi hingga 626.155 pon CO2e. Angka ini setara dengan hampir lima
kali lipat total polusi yang dihasilkan sebuah mobil selama seumur hidupnya.
- Ketimpangan Akses Riset: Tingginya biaya listrik dan perangkat keras (hardware) ini membuat riset AI mutakhir hanya bisa didominasi oleh perusahaan besar, sementara peneliti akademis kesulitan bersaing karena keterbatasan dana.

> Cara Mengatasi (Solusi):
- Transparansi Pelaporan: Para peneliti AI wajib mencantumkan waktu pelatihan dan
tingkat konsumsi daya di dalam jurnal mereka agar publik bisa menilai apakah model
tersebut efisien dan sepadan dengan energi yang disedot.
- Metode yang Cerdas dan Efisien: Mengganti cara pelatihan lama (grid search) yang menguji sistem secara buta dan boros energi, dengan metode algoritma cerdas (seperti pencarian acak atau Bayesian) yang lebih hemat komputasi.
- Pusat Komputasi Bersama: Pemerintah disarankan membangun fasilitas komputasi
cloud bersama khusus untuk akademisi. Hal ini akan mencegah pemborosan energi
dari pembelian alat yang berlebihan dan membuat akses riset lebih adil.
- Pendapat Pribadi: Para pengembang harus lebih mengutamakan pendekatan transfer learning. Daripada menghabiskan listrik masif untuk melatih model AI dari nol, lebih baik menggunakan model yang sudah dilatih (pre trained model) dan hanya memodifikasi lapisan ujungnya saja (fine tuning) untuk tugas spesifik. Langkah ini menghemat waktu, biaya, dan jejak karbon secara signifikan.
